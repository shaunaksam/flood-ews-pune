```python
import pandas as pd

df = pd.read_csv("../data/raw/drvineet_2014_2026.csv", header=None)

clean_data = []

for i in range(0, len(df.columns), 3):

    # First column = dates, second column = rainfall
    date_col = df.columns[i]
    value_col = df.columns[i + 1]

    # Year is stored in row 2 of the spreadsheet
    year = df.iloc[1, i]

    # Start from row 3 of the spreadsheet
    for row in range(2, len(df)):

        date_value = df.iloc[row, i]
        rainfall = df.iloc[row, i + 1]

        # Skip rows where the date is missing
        if pd.isna(date_value):
            continue

        # Extract month and day, ignore the year in the date string
        parts = date_value.split("/")

        # Rebuild the date using the year from the block
        date_string = parts[0] + "/" + parts[1] + "/" + str(year)

        # Convert to a real datetime
        date = pd.to_datetime(date_string, format="%m/%d/%Y")

        # Add one clean record
        clean_data.append({
            "date": date,
            "rainfall": rainfall
        })

# Create the new long-format DataFrame
clean_df = pd.DataFrame(clean_data)

# Sort chronologically
clean_df = clean_df.sort_values("date").reset_index(drop=True)

print(clean_df)
```

```python
print(clean_df["rainfall"].isna().sum())
```

```python
clean_df = clean_df[clean_df["date"] <= "2026-08-30"]
```
