```python
import pandas as pd
wards = pd.read_csv("../data/processed/ward_features_v1.csv")
```

```python
elev_min = wards["elevation_mean"].min()
elev_max = wards["elevation_mean"].max()
wards["elevation_risk"] = (wards["elevation_mean"] - elev_max) / (elev_min - elev_max)

inc_min = wards["dam_discharge_incident_count"].min()
inc_max = wards["dam_discharge_incident_count"].max()
wards["incident_risk"] = (wards["dam_discharge_incident_count"] - inc_min) / (inc_max - inc_min)

w1, w2 = 0.4, 0.6
wards["risk_score"] = w1 * wards["elevation_risk"] + w2 * wards["incident_risk"]

print(wards.sort_values("risk_score", ascending=False).head(10)[["Name2", "elevation_mean", "dam_discharge_incident_count", "risk_score"]])
```

```python
wards.to_csv("../data/processed/ward_risk_v1.csv", index=False)
```
